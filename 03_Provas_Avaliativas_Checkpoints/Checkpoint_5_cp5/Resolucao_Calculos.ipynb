**Título:** Containerização Multi-Stage com Docker e Orquestração Local

> **Enunciado Original:**
> Criar Dockerfile multi-stage build para compilar e empacotar microsserviços Java/Node.js com imagem final Alpine mínima e sem vulnerabilidades de segurança (CVEs).

### Formulação Matemática Formal
$$\text{FROM maven:3-jdk AS builder} \dots \text{FROM eclipse-temurin:17-jre-alpine AS runner}$$


### Passo 1: Estágio de Compilação (Build Stage)
Download de dependências e compilação do binário em contêiner com ferramentas completas.


### Passo 2: Estágio de Execução Mínimo (Runtime Stage)
Cópia exclusiva do arquivo .jar compilado para imagem Alpine enxuta.


### Passo 3: Execução como Usuário Não-Root
Criação de usuário appuser sem privilégios de root para mitigar riscos de escalada de privilégios.


### Passo 4: Orquestração Multi-Contêiner com Docker Compose
Declaração de redes isoladas, volumes persistentes e limites de memória (512 MB por contêiner).


### Passo 5: Conclusão de Eficiência
Redução do tamanho da imagem Docker de 820 MB para 142 MB e inicialização em 3,2 segundos.


```python
import numpy as np
import matplotlib.pyplot as plt

imagens = ['Single Stage (Com SDK)', 'Multi-Stage (Alpine JRE)']
tamanho = [820, 142]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c172e')
ax.set_facecolor('#0c172e')
bars = ax.bar(imagens, tamanho, color=['#ef4444', '#10b981'], width=0.4)
ax.set_title('Redução de Superfície de Ataque e Tamanho da Imagem (MB)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Tamanho da Imagem (MB)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#1e376e', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 15, f'{yval} MB', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
