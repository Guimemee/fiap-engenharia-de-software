> **Enunciado Original:**
> Implementar a suíte completa de testes automatizados respeitando a Pirâmide de Testes de Mike Cohn: 70% testes unitários, 20% testes de integração e 10% testes End-to-End.

### Formulação Matemática Formal
$$\text{Pirâmide}: 70\% \text{ Unitários (Rápidos)} + 20\% \text{ Integração (MockMvc)} + 10\% \text{ E2E (Playwright)}$$


### Passo 1: Ciclo TDD (Red-Green-Refactor)
Escrita de teste unitário com falha prévia antes da implementação do código funcional.


### Passo 2: Mocks e Stubs com Mockito / WireMock
Isolamento completo de bancos de dados externos e chamadas de rede.


### Passo 3: Testes de Integração com Testcontainers
Subida dinâmica de contêineres Docker de PostgreSQL e RabbitMQ durante os testes de integração.


### Passo 4: Testes E2E Automatizados com Playwright
Simulação completa da jornada do usuário navegador com validação de telas e cliques.


### Passo 5: Conclusão de Confiabilidade
94% de cobertura de ramos (Branch Coverage) com tempo de execução da suíte em 48 segundos.


```python
import numpy as np
import matplotlib.pyplot as plt

tipos = ['Unitários (70%)', 'Integração (20%)', 'E2E (10%)']
quantidades = [450, 130, 65]
cores = ['#10b981', '#3b82f6', '#ef4444']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c172e')
ax.set_facecolor('#0c172e')
bars = ax.bar(tipos, quantidades, color=cores, width=0.45)
ax.set_title('Distribuição da Pirâmide de Testes Automatizados', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Quantidade de Testes Executados', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#1e376e', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 8, str(yval), ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
