> Implementar os padrões Factory Method, Observer e Strategy para construir um pipeline extensível de processamento de alertas e notificações industriais.

### Formulação Matemática Formal
$$\text{Observer Pattern}: \text{Subject.notifyObservers}() \implies \forall o \in \text{Observers}: o.\text{update}(event)$$


### Passo 1: Padrão Factory Method
Criação desacoplada de notificações (EmailNotifier, SMSNotifier, WebhookNotifier) via interface comum.


### Passo 2: Padrão Observer para Publicação/Assinatura
O SensorSubject notifica múltiplos ouvintes assíncronos sem acoplamento direto.


### Passo 3: Padrão Strategy para Algoritmos de Triagem
Intercâmbio dinâmico entre algoritmos de detecção de anomalias (Limiar Fixo, Média Móvel, Isolation Forest).


### Passo 4: Princípios SOLID Validados
Open/Closed Principle (aberto para extensão, fechado para modificação) e Single Responsibility.


### Passo 5: Conclusão de Arquitetura
Adição de novos canais de alerta sem alterar nenhuma linha do código de sensoriamento central.


```python
import numpy as np
import matplotlib.pyplot as plt

metricas = ['Acoplamento Aferente (Ca)', 'Acoplamento Eferente (Ce)', 'Instabilidade (I = Ce/(Ca+Ce))']
valores = [8, 1, 0.11]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c172e')
ax.set_facecolor('#0c172e')
bars = ax.barh(metricas, valores, color='#3b82f6', height=0.4)
ax.set_title('Métricas de Acoplamento Modular de Martin (GoF)', color='#ffffff', fontsize=11, fontweight='bold')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='x', color='#1e376e', alpha=0.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
