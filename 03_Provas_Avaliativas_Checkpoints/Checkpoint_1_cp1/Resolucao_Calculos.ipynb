> **Enunciado Original:**
> Mapear os requisitos funcionais e não-funcionais do sistema de telemetria hospitalar Sanofi, construindo épicos, User Stories investigáveis e cenários BDD (Given-When-Then).

### Formulação Matemática Formal
$$\text{User Story}: \text{Como } [\text{Persona}] \text{ Eu quero } [\text{Ação}] \text{ Para que } [\text{Valor de Negócio}]$$


### Passo 1: Elicitação com Stakeholders
Mapeamento de dores com médicos, químicos industriais e gestores de produção.


### Passo 2: Definição dos Requisitos Não-Funcionais (RNFs)
Disponibilidade 99.95%, tempo de resposta P95 < 200 ms e conformidade com LGPD e FDA 21 CFR Part 11.


### Passo 3: Redação de Histórias no Padrão INVEST
Independente, Negociável, Valiosa, Estimável, Pequena (Small) e Testável.


### Passo 4: Cenários de Aceite com Cucumber Gherkin
Given sensor emite alerta térmico, When temperatura > 80 °C, Then disparar parada preventiva.


### Passo 5: Conclusão do Backlog
Product Backlog priorizado por valor de negócio e matriz MoSCoW.


```python
import numpy as np
import matplotlib.pyplot as plt

prioridades = ['Must Have (Essencial)', 'Should Have (Importante)', 'Could Have (Desejável)', 'Won't Have (Futuro)']
contagem = [18, 12, 8, 4]
cores = ['#ef4444', '#3b82f6', '#10b981', '#94a3b8']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c172e')
ax.set_facecolor('#0c172e')
bars = ax.bar(prioridades, contagem, color=cores, width=0.5)
ax.set_title('Priorização de Requisitos pelo Método MoSCoW', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Quantidade de User Stories', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#1e376e', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 0.4, str(yval), ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
