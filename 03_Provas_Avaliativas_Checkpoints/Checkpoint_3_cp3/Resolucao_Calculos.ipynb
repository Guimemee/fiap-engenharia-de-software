**Título:** Arquitetura de Microsserviços e Padrão API Gateway

> **Enunciado Original:**
> Decompor a aplicação monolítica em microsserviços autônomos orientados a domínio (DDD) com roteamento centralizado via API Gateway e mensageria RabbitMQ.

### Formulação Matemática Formal
$$\text{Clients} \to \text{API Gateway} \xrightarrow{\text{Service Discovery}} \{\text{Telemetry-Svc}, \text{Auth-Svc}, \text{Report-Svc}\}$$


### Passo 1: Identificação de Bounded Contexts com DDD
Separação de domínios: Ingestão de Sensores, Gestão de Usuários e Auditoria Regulatória.


### Passo 2: Padrão API Gateway
Centralização de autenticação JWT, rate limiting (100 req/s por cliente) e terminação TLS.


### Passo 3: Comunicação Assíncrona com RabbitMQ
Filas com Dead Letter Exchange (DLX) para mensagens rejeitadas ou com timeout.


### Passo 4: Padrão Circuit Breaker (Resilience4j)
Abertura de circuito após 5 falhas consecutivas, prevenindo falha em cascata (Cascading Failure).


### Passo 5: Conclusão de Disponibilidade
Queda de um microsserviço secundário não impacta a operação contínua da linha de produção.


```python
import numpy as np
import matplotlib.pyplot as plt

servicos = ['API Gateway', 'Telemetry Svc', 'Auth Svc', 'Audit Svc', 'RabbitMQ']
latencia_p95 = [12, 45, 18, 62, 8]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c172e')
ax.set_facecolor('#0c172e')
bars = ax.bar(servicos, latencia_p95, color='#10b981', width=0.45)
ax.set_title('Latência P95 por Microsserviço (ms)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Latência de Resposta P95 (ms)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#1e376e', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 1, f'{yval}ms', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
