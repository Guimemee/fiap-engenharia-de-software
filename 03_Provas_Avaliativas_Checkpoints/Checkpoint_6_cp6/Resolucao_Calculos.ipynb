**Título:** Pipeline de Integração e Entrega Contínua (CI/CD) com GitHub Actions

> **Enunciado Original:**
> Projetar workflow automatizado de CI/CD no GitHub Actions contemplando linting, testes, análise estática SonarQube, build de imagem Docker e deploy automatizado no Kubernetes.

### Formulação Matemática Formal
$$\text{Commit} \to \text{Lint} \to \text{Test} \to \text{SonarQube} \to \text{Docker Push} \to \text{K8s Rolling Update}$$


### Passo 1: Disparo Automático por Push e PR
Gatilhos on: push: branches: [main] e pull_request para branches de feature.


### Passo 2: Quality Gate com SonarCloud
Bloqueio da esteira caso cobertura < 80% ou existência de qualquer vulnerabilidade crítica.


### Passo 3: Publicação em Container Registry
Geração de tags semânticas (v1.2.3 e commit SHA) e push no GitHub Packages.


### Passo 4: Estratégia de Deploy Rolling Update no Kubernetes
Atualização contínua com maxSurge=1 e maxUnavailable=0 garantindo zero downtime.


### Passo 5: Conclusão DevOps
Tempo médio de ciclo de entrega (Lead Time for Changes) reduzido de 2 dias para 8 minutos.


```python
import numpy as np
import matplotlib.pyplot as plt

etapas = ['Lint & Check', 'Testes Unitários', 'SonarQube Gate', 'Docker Build', 'Deploy K8s']
duracao_s = [25, 48, 42, 75, 30]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c172e')
ax.set_facecolor('#0c172e')
bars = ax.barh(etapas, duracao_s, color='#3b82f6', height=0.45)
ax.set_title('Tempo de Execução do Pipeline CI/CD GitHub Actions (s)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo em Segundos', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='x', color='#1e376e', alpha=0.5)
for bar in bars:
    wval = bar.get_width()
    ax.text(wval + 2, bar.get_y() + bar.get_height()/2.0, f'{wval}s', ha='left', va='center', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
